```python
# 2. dataclass：实现 Document / AuditResult，AuditResult.reasons 必须正确使用 default_factory。
from dataclasses import dataclass, field

@dataclass(kw_only=True)
class Document:
    id: int
    title: str
    author: str
    price: float

@dataclass(kw_only=True)
class AuditResult:
    document_id: int
    audit_admin_id: int
    audit_time: int
    audit_status: int
    reasons: list[str] = field(default_factory=list)
```

```python
# 3. Pydantic：实现 DocumentCreate，并准备至少 8 组输入数据验证。
from pydantic import BaseModel, Field, ValidationError
import pytest

class DocumentCreate(BaseModel):
    title: str = Field(min_length=2, max_length=20)
    author: str
    price: float = Field(ge=0.01, le=50)

dc1 = DocumentCreate(title="商品", author="Conley", price=0.01)
assert dc1.model_dump() == {'title': '商品', 'author': 'Conley', 'price': 0.01}
print("数据验证1：最小边界", dc1.model_dump())
print(40 * "-")
dc2 = DocumentCreate(title="商品2商品2商品2商品2商品2商品商品2", author="Conley", price=50)
assert dc2.model_dump() == {'title': '商品2商品2商品2商品2商品2商品商品2', 'author': 'Conley', 'price': 50.0}
print("数据验证2：最大边界", dc2.model_dump())
print(40 * "-")
dc3 = DocumentCreate(title="五年级下册美术说课稿", author="renrendoc", price=12.99)
assert dc3.model_dump() == {'title': '五年级下册美术说课稿', 'author': 'renrendoc', 'price': 12.99}
print("数据验证3：正常数据", dc3.model_dump())
print(40 * "-")
with pytest.raises(ValidationError) as exc_info:
    dc4 = DocumentCreate(title="商", author="Conley", price=0.01)
assert "at least 2 characters" in str(exc_info.value)
print("数据验证4：标题不能小于2个字", exc_info)
print(40 * "-")
with pytest.raises(ValidationError) as exc_info:
    dc5 = DocumentCreate(title="商品5", price=0.01)
assert "Field required [type=missing" in str(exc_info.value)
print("数据验证5：author必传", exc_info)
print(40 * "-")
with pytest.raises(ValidationError) as exc_info:
    dc6 = DocumentCreate(title="商品5", author="Conley", price=0.00)
assert "be greater than or equal to 0.01" in str(exc_info.value)
print("数据验证6：price不能小于0.01", exc_info)
print(40 * "-")
with pytest.raises(ValidationError) as exc_info:
    dc7 = DocumentCreate(title="商品5", author="Conley", price=50.01)
assert "be less than or equal to 50" in str(exc_info.value)
print("数据验证7：price不能大于50", exc_info)
print(40 * "-")
with pytest.raises(ValidationError) as exc_info:
    dc8 = DocumentCreate(title="商品2商品2商品2商品2商品2商品2商品2", author="Conley", price=0.01)
assert "at most 20 characters" in str(exc_info.value)
print("数据验证8：标题不能大于20个字", exc_info)
print(40 * "-")
```
